```python
from pydjirecord import DJILog
from pydjirecord.frame.details import FrameDetails
import json

#using https://github.com/rembish/pydjirecord

filename = "FlightRecord_2025-08-09_[15-59-59].txt"
data = open(f"data/input/{filename}", "rb").read()
log = DJILog.from_bytes(data)

with open("keys.json") as f:
        data = json.load(f)

dji_key = data["DJI_API"]
```

```python
#Decrypt 
keychains = log.fetch_keychains(dji_key) if log.version >= 13 else None

#get frames
frames = log.frames(keychains)

# FrameDetails computes all corrected values from frames automatically
details = FrameDetails.from_details(log.details, frames)

records = log.records(keychains)
```

```python
for record in records:
    print(record)
```
